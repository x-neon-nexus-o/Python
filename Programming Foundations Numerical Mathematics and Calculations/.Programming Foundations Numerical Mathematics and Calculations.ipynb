# Writing a program to calculate factors of an integer

```python
def calculate_factors(number):
    list_of_factors = []
    for i in range(1,number+1):
        if number % i ==0:
            list_of_factors.append(i)
    return list_of_factors
```

```python
calculate_factors(30)
```

# Challenge: Converting units of measurement



```python
def select_conversion():
    print("Select conversion type:")
    print("1.Kilometers to Miles")
    print("2.Miles to Kilometers")
    print("0.Exit")
    
def kilometers_to_miles():
    km=float(input("Enter the distance in kilometers: "))
    miles=km*0.621371
    print(f"{km} kilometers is equal to {miles} miles")
    
def miles_to_kilometers():
    miles=float(input("Enter the distance in miles: "))
    km=miles/0.621371
    print(f"{miles} miles is equal to {km} kilometers")

select_conversion()
option=input("Enter option (1/2/0): ")
if option=='1':
    kilometers_to_miles()
elif option=='2':
    miles_to_kilometers()
elif option=='0':
    print("Exiting the program.")
else:
    print("Invalid option. Please select a valid option.")
```

# Transposing matrices



```python
def transpose(a):
    matrix=[]
    m=len(a)
    n=len(a[0])
    for i in range(n):
        matrix.append([])
        for j in range(m):
            matrix[i].append(a[j][i])
    return matrix
```

```python
a=[[1,5,6],[4,8,3]]
transpose(a)
```
