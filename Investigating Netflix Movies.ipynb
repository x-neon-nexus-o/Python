```python
# Importing pandas and matplotlib
import pandas as pd
import matplotlib.pyplot as plt

# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("D:/Python/netflix_titles.csv")
netflix_df.head()
```

```python
netflix_subset=netflix_df[netflix_df["type"]=="Movie"]
```

```python
subset=netflix_subset[(netflix_subset["release_year"]>=1990)]
```

```python
movies_1990s = subset[(subset["release_year"] < 2000)]
```

```python
plt.hist(movies_1990s["duration"])
plt.title("Distribution of Movie Durations")
plt.xlabel("Duration (minutes)")
plt.ylabel("Number of Movies")
plt.show()
```

```python
duration=100

def contains_string(text, keyword):
    return isinstance(text, str) and keyword.lower() in text.lower()

action_movies = movies_1990s[movies_1990s["listed_in"].apply(lambda x: contains_string(x, "Action"))].copy()
action_movies["duration"] = action_movies["duration"].str.extract(r"(\d+)").astype(int)
```

```python
short_movie_count=0

for label,row in action_movies.iterrows():
    if row["duration"]<90:
        short_movie_count=short_movie_count+1
    else:
        short_movie_count=short_movie_count
print(short_movie_count)
```
